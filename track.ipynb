```python
import cv2
import numpy as np
from ultralytics import YOLO
from boxmot import ByteTrack, BotSort, DeepOcSort, OcSort, StrongSort
from cute_vision.tasks.track import result_to_boxmot_dets
```

```python
tracker = ByteTrack(
    track_thresh=0.25,
    track_buffer=30,
    match_thresh=0.8,
    frame_rate=30
)
```

```python
frame = cv2.imread('humans.png')
frame2 = cv2.imread('humans2.png')
```

```python
model = YOLO('yolov8n-pose.pt')
```

```python
results = model(frame, classes=[0], conf=0.6, verbose=False)
result = results[0]
dets1 = result_to_boxmot_dets(result)
```

```python
results = model(frame2, classes=[0], conf=0.6, verbose=False)
result = results[0]
dets2 = result_to_boxmot_dets(result)
```

```python
tracker.update(dets1, frame)
```

```python
tracker.update(dets2, frame2)
```

```python
image = tracker.plot_results(frame2, show_trajectories=True)
```

```python
cv2.imwrite('output.png', image)
```
